```python
import pandas as pd
```

```python
#Load the CSV file using pandas.
df=pd.read_csv("students.csv")
```

```python
#Display the first 5 rows.
df.head()
```

```python
#How many students are there in total?
df["Student_ID"].count()
```

```python
#Show only Data Science students.
df[df["Department"]=="Data Science"]
```

```python
#Find all students who failed.
df[df["Result"]=="Fail"]
```

```python
#Find the average attendance of all students.
df["Attendance_Percent"].mean()
```

```python
#Find the average Python mark department-wise.
df.groupby("Department")["Python"].mean().reset_index()
```

```python
#Which student has the highest study hours?
high=df.iloc[df["Study_Hours"].idxmax()]
high.reset_index()
```

```python
#Count how many students passed and failed.
pass_list=df[df["Result"]=="Pass"].count()
Fail_list=df[df["Result"]=="Fail"].count()
print(pass_list[["Student_ID"]])
print(Fail_list[["Student_ID"]])
```

```python
#Find students with attendance less than 60%.
df[df["Attendance_Percent"]<60]
```

```python
#Add a new column Total_Marks (Maths + Python + Statistics).
df["Total_Marks"]=df["Maths"]+df["Python"]+df["Statistics"]
df
```

```python
#Add Average_Marks column.
df["Average_Marks"]=(df["Python"]+df["Maths"]+df["Statistics"]/3)
df
```

```python
#Find students whose: Attendance > 75 AND Average_Marks > 70
df[(df["Attendance_Percent"]>75)&(df["Average_Marks"]>70)]
```

```python
#Which department has the highest pass percentage?
total=df.groupby("Department")["Student_ID"].count()
passed=df[df["Result"]=="Pass"].groupby("Department")["Student_ID"].count()
pass_percentage=(passed/total)*100
highest_pass_dept=pass_percentage.idxmax()
print("Highest Pass Percentage:")
print(highest_pass_dept)
print(f"Highest Percentage:{pass_percentage[highest_pass_dept]}")
```

```python
#Check if study hours really affect results (compare avg study hours of Pass vs Fail).
avg_study_hours=df.groupby("Result")["Study_Hours"].mean()
avg_study_hours
```

```python
#List students who are at risk: Attendance < 60 OR Study_Hours < 1
df[(df["Attendance_Percent"]<60) | (df["Study_Hours"]<1)]
```

```python
#Sort students by Python marks (descending).
df.sort_values("Python",ascending=False)
```

```python
#Find top 3 performers based on Total_Marks.
df.sort_values("Total_Marks",ascending=False).head(3)
```

```python
#Replace missing values (if any) with column average.
print(df.isna().sum())
df.fillna(df.mean(numeric_only=True),inplace=True)
print(df.isna().sum())
```

```python
#Export only passed students to a new CSV file.
passed_students=df["Result"]=="Pass"
passed_students.to_csv("passed_students.csv",index=False)
print('passed_students.scv exported successfully')
```
